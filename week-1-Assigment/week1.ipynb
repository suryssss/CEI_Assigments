Handling Missing Values

```python
#handling missing values
#missing values in each column
df.isnull().sum()
```

```python
'''
the columns having null values are discount,what_customers_said, 
seller_name,videos ,seller_information and variantions
%of missing values of discount =12.1%
%of missing values of what_customers_said= 57.3%
%of missing values of seller_name=30.1%
%of missing values of videos=78.1%
%of missing values of seller_infomation=30.1%
%of missing values of variations=56.2
 '''

#filling null data
#filling discount with median
df["discount"] = df["discount"].fillna(df["discount"].median())
```

```python
#seller_columns non null values
print(df["seller_name"].sample(10))
```

```python
#handling missing value of seller_name
df["seller_name"] = df["seller_name"].fillna("Unknown Seller")
```

```python
#displaying seller_information column
print(df["seller_information"].sample(10))
```

```python
#handling missing values for seller_information
df["seller_information"] = df["seller_information"].fillna("Information not Found")
```

```python
#display sample data of what_customer_said column
print(df['what_customers_said'].dtype) #object

#example data 
print(df[df["what_customers_said"].notna()]["what_customers_said"])
```

```python
#despite having 50%+ null values as we cannot ignore the feedback from the customers as they are 
#core business logic 
#handling the null values by adding an indicator flag (feature engineering !!)

#creating a indicator column as has_customer_reviews
df["has_customer_reviews"]=df['what_customers_said'].notna().astype(int)

#filling the null values with an empty list to preserve sturcture
df["what_customers_said"] = df["what_customers_said"].fillna("[]")

# Verify
print(df["what_customers_said"].isnull().sum())

#view the result
df[['what_customers_said', 'has_customer_reviews']].sample(10)
```

```python
#sample data of vide0s and variations column which are not null
print(df["videos"].sample(10))
```

```python
#dropping videos column as doesnt effect the actual analysis of the data
df=df.drop(columns=["videos"])

#verification 
print(df.columns)
```

```python
#sample data of variations column
print(df["variations"].sample(10))
```

```python
#dropping the variations column as there is no valid data
df=df.drop(columns=["variations"])

print(df.columns)
```
