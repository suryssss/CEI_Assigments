 Handle null values and filter .

```python
from pyspark.sql.functions import count, when
#counting null values in all the columns
df.select([
    count(when(col(c).isNull(), c)).alias(c)
    for c in df.columns
]).show()

```

```python
#handling email column

from pyspark.sql.functions import lit

df = df.na.fill({"email": "Not Provided"})
```

```python
df.filter(col("email").isNull()).count()
```

```python
from pyspark.sql.functions import col

df.filter(col("quantity").isNull()) \
  .select(
      "order_id",
      "quantity",
      "unit_price",
      "total_amount"
  ) \
  .show(truncate=False)
```

```python
from pyspark.sql.functions import col

df.filter(
    col("quantity").isNull() &
    (
        col("unit_price").isNotNull() |
        col("total_amount").isNotNull()
    )
).count()


#either of unit_price or total_amount is being null to fill the quantity using the formula quantity=total_amount/unit_price
#so cant fill them just dropping 
```

```python
#dropping the null values
df=df.na.drop(subset=["quantity"])
```

```python
df.filter(col("quantity").isNull()).count()
```

```python
df.filter(col("discount_price").isNull()).count()
```

```python
df.filter(col("total_amount").isNull()).count()
```

```python
from pyspark.sql.functions import col, when

#handling discount percent by calculating from discount_price and unit_price

df = df.withColumn(
    "discount_percent",
    when(
        col("discount_percent").isNull() &
        col("unit_price").isNotNull() &
        col("discount_price").isNotNull(),

        (1 - col("discount_price") / col("unit_price")) * 100
    ).otherwise(col("discount_percent"))
)
```

```python
df.filter(col("discount_percent").isNull()).count()
```

```python
 # payment method 
 df.select("payment_method").distinct().show(truncate=False)
```

```python
#filling the null values of payment method with unknown
df= df.na.fill({"payment_method": "Unknown"})
```

```python
df.filter(col("payment_method").isNull()).count()
```

```python
df.select("rating").distinct().show()
```

```python
# in rating the null could be business value as the customer didnt want to leave a rating hence i could leave as it is 
# but if i need to remove null values filling it with average rating would be a good idea

from pyspark.sql.functions import avg
avg_rating = float(round(df.select(avg("rating")).first()[0]))
df = df.na.fill({"rating": avg_rating})
```

```python
df.filter(col("rating").isNull()).count()
```

```python
from pyspark.sql.functions import count, when
#counting null values in all the columns
df.select([
    count(when(col(c).isNull(), c)).alias(c)
    for c in df.columns
]).show()

#handled all null values
```
